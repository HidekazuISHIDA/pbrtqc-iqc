```python
import shutil, threading, http.server, functools
shutil.copy(ROOT / "monitor" / "index.html", C.MONITOR / "index.html")
PORT = 8791
handler = functools.partial(http.server.SimpleHTTPRequestHandler, directory=str(C.MONITOR))
srv = http.server.ThreadingHTTPServer(("127.0.0.1", PORT), handler)
threading.Thread(target=srv.serve_forever, daemon=True).start()
print(f"ブラウザで開く: http://127.0.0.1:{PORT}/?t=2025-10-02T10:00&item=Ca   （止めるにはカーネルを再起動）")
```
